```python
def analyze_text(sentence: str) -> dict:
    vowels = "aeiou"
    sentence = sentence.lower()
    vowel_count = sum(1 for char in sentence if char in vowels)
    consonant_count = sum(1 for char in sentence if char.isalpha() and char not in vowels)
    word_count = len(sentence.split())
    sentence_length = len(sentence)
    digit_count = sum(1 for char in sentence if char.isdigit())
    special_char_count = sum(1 for char in sentence if not char.isalnum() and not char.isspace())
    return {
        "vowel_count": vowel_count,
        "consonant_count": consonant_count,
        "word_count": word_count,
        "sentence_length": sentence_length,
        "digit_count": digit_count,
        "special_char_count": special_char_count
    }
```

```python
# Create a DataFrame with duplicate and missing values
data = {
    'ID': [1, 2, 3, 4, 5, 5, 6, None],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Eve', None, 'Grace'],
    'Age': [25, 30, 35, None, 40, 40, 45, 50],
    'Gender': ['F', 'M', 'M', 'M', 'F', 'F', None, 'F']
}

data = pd.DataFrame(data)
print(data)
```

```python
data.dropna(inplace=True)  # Drop rows with any missing values
print("\nData after dropping rows with missing values:")
print(data)
```

```python
data.groupby('Gender')['Age'].mean()
```

```python
print("\nData after sorting by Age in descending order:")
print(data.sort_values(by='Name', ascending=True))
```

```python
data.drop_duplicates(subset=['ID'], keep='first', inplace=True)  # Drop duplicate IDs, keeping the first occurrence
```

```python
import pandas as pd

data = {
    'Department': ['Sales', 'Sales', 'HR', 'HR', 'IT', 'IT'],
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank'],
    'Salary': [50000, 55000, 60000, 62000, 70000, 72000],
    'Bonus': [5000, 7000, 6000, 6500, 8000, 8500]
}

df = pd.DataFrame(data)
df
```

```python
df.groupby(['Department','Employee']).agg({'Salary': 'sum', 'Bonus': 'mean'}).reset_index()
```

```python
df.groupby('Department').agg({'Salary': ['mean', 'max'], 'Bonus': ['sum', 'mean']}).reset_index()
```
