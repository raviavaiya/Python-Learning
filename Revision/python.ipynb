```python
# int()
num_str = "123"
num_int = int(num_str)  # num_int will be 123 (integer)
print(f"int({num_str}) = {num_int}, type: {type(num_int)}")
```

```python
# float()
int_num = 10
float_num = float(int_num) # float_num will be 10.0 (float)
print(f"float({int_num}) = {float_num}, type: {type(float_num)}")
```

```python
str_float = "3.14"
actual_float = float(str_float) # actual_float will be 3.14 (float)
print(f"float({str_float}) = {actual_float}, type: {type(actual_float)}")
```

```python
# str()
num = 123
str_num = str(num) # str_num will be "123" (string)
print(f"str({num}) = '{str_num}', type: {type(str_num)}")
```

```python
# bool()
# Empty strings, empty lists, empty tuples, empty dictionaries, empty sets, and 0 are considered False.
# Anything else is True.
print(f"bool(0) = {bool(0)}, bool(1) = {bool(1)}")
print(f"bool('') = {bool('')}, bool('hello') = {bool('hello')}")
print(f"bool([]) = {bool([])}, bool([1,2]) = {bool([1,2])}")
print(f"bool({{}}) = {bool({})}, bool({{'a':1}}) = {bool({'a':1})}")
```

```python
list1 = [1, 2, 3]
list2 = list1 # list2 now refers to the *same* list object as list1

print(f"Before modification:")
print(f"list1: {list1}, id(list1): {id(list1)}")
print(f"list2: {list2}, id(list2): {id(list2)}")

list1.append(4) # Modify list1

print(f"\nAfter modification:")
print(f"list1: {list1}, id(list1): {id(list1)}")
print(f"list2: {list2}, id(list2): {id(list2)}") # list2 also changed!

# Note that the memory address (id) remains the same, indicating it's the same object being modified.
```

```python
str1 = "hello"
str2 = str1 # str2 refers to the same string object as str1

print(f"Before 'modification':")
print(f"str1: {str1}, id(str1): {id(str1)}")
print(f"str2: {str2}, id(str2): {id(str2)}")

str1 = str1 + " world" # This operation *creates a new string* for str1

print(f"\nAfter 'modification':")
print(f"str1: {str1}, id(str1): {id(str1)}")
print(f"str2: {str2}, id(str2): {id(str2)}") # str2 remains unchanged

# Notice that the id of str1 changed, meaning a new object was created.
# The original "hello" string still exists in memory and is referred to by str2.
```

## Function Arguments

### positional arguments

```python
def describe_pet(animal_type, pet_name):
    """Displays information about a pet."""
    print(f"\nI have a {animal_type}.")
    print(f"My {animal_type}'s name is {pet_name}.")

# Correct usage: position matters
describe_pet('hamster', 'Harry') # animal_type is 'hamster', pet_name is 'Harry'
describe_pet('dog', 'Willie')   # animal_type is 'dog', pet_name is 'Willie'

# Incorrect usage (missing argument):
# describe_pet('cat') # TypeError: describe_pet() missing 1 required positional argument: 'pet_name'

# Incorrect usage (extra argument):
# describe_pet('fish', 'Nemo', 'small') # TypeError: describe_pet() takes 2 positional arguments but 3 were given
```

###  Default Arguments (Optional Parameters)

```python
def describe_car(make, model, color="silver"): # 'color' has a default value
    """Displays information about a car."""
    print(f"\nMake: {make}")
    print(f"Model: {model}")
    print(f"Color: {color}")

# Using the default color
describe_car("Toyota", "Camry")
# Output:
# Make: Toyota
# Model: Camry
# Color: silver

# Overriding the default color
describe_car("Honda", "Civic", "red")
# Output:
# Make: Honda
# Model: Civic
# Color: red

# Important note on mutable default arguments:
# Be careful when using mutable objects (like lists or dictionaries) as default arguments.
# They are initialized only once when the function is defined, not on each call.
def add_item(item, my_list=[]): # DANGER: my_list is created once
    my_list.append(item)
    return my_list

print(add_item('apple'))  # ['apple']
print(add_item('banana')) # ['apple', 'banana'] - unexpected!
print(add_item('orange', ['custom_list'])) # ['custom_list', 'orange'] - works as expected here

# Correct way to handle mutable defaults:
def add_item_safe(item, my_list=None):
    if my_list is None:
        my_list = []
    my_list.append(item)
    return my_list

print(add_item_safe('apple'))  # ['apple']
print(add_item_safe('banana')) # ['banana'] - now works as expected
```

### Keyword Arguments


```python
def create_user(username, password, email):
    """Creates a new user profile."""
    print(f"Username: {username}")
    print(f"Password: {password}")
    print(f"Email: {email}")

# Using positional arguments (order matters)
create_user("john_doe", "pass123", "john@example.com")

# Using keyword arguments (order doesn't matter, readability improved)
create_user(email="jane@example.com", password="securepwd", username="jane_smith")

# Mixing positional and keyword arguments:
# Positional arguments must come before keyword arguments.
def order_pizza(size, toppings, cheese=True):
    print(f"Ordering a {size} pizza with {toppings}")
    if cheese:
        print("With extra cheese!")
    else:
        print("No extra cheese.")

order_pizza("large", "pepperoni", cheese=False) # Positional first, then keyword
# order_pizza(cheese=True, "medium", "mushrooms") # SyntaxError: positional argument follows keyword argument
```

### Arbitrary Positional Arguments, *args

```python
def sum_all_numbers(*numbers):
    """Calculates the sum of an arbitrary number of integers."""
    total = 0
    for num in numbers:
        total += num
    return total

print(sum_all_numbers(1, 2, 3))         # Output: 6
print(sum_all_numbers(10, 20, 30, 40)) # Output: 100
print(sum_all_numbers())               # Output: 0 (empty tuple)

def print_items(first_item, *remaining_items):
    print(f"First item: {first_item}")
    print(f"Remaining items (tuple): {remaining_items}")

print_items("apple", "banana", "cherry", "date")
# Output:
# First item: apple
# Remaining items (tuple): ('banana', 'cherry', 'date')
```

### Arbitrary Keyword Arguments (**kwargs)

```python
def build_profile(first, last, **user_info):
    """Builds a dictionary containing everything we know about a user."""
    user_info['first_name'] = first
    user_info['last_name'] = last
    return user_info

profile1 = build_profile('albert', 'einstein', location='princeton', field='physics')
print(profile1)
# Output: {'location': 'princeton', 'field': 'physics', 'first_name': 'albert', 'last_name': 'einstein'}

profile2 = build_profile('marie', 'curie', country='france', profession='scientist', discovery='radioactivity')
print(profile2)
# Output: {'country': 'france', 'profession': 'scientist', 'discovery': 'radioactivity', 'first_name': 'marie', 'last_name': 'curie'}

def display_config(**config):
    # for key, value in config.items():
    #     print(f"{key}: {value}")
    return config

display_config(server="production", port=8080, debug_mode=False)
# Output:
# server: production
# port: 8080
# debug_mode: False
```

```python
# range(stop)
for i in range(5): # Generates 0, 1, 2, 3, 4
    print(i)

print("-" * 10)

# range(start, stop)
for i in range(2, 7): # Generates 2, 3, 4, 5, 6
    print(i)

print("-" * 10)

# range(start, stop, step)
for i in range(1, 10, 2): # Generates 1, 3, 5, 7, 9
    print(i)

print("-" * 10)

# Counting downwards
for i in range(5, 0, -1): # Generates 5, 4, 3, 2, 1
    print(i)
```

```python
class Dog:
    # Class attribute (shared by all instances of Dog)
    species = "Canis familiaris"

    def __init__(self, name, age):
        """
        The constructor method.
        It's called automatically when a new object of the class is created.
        'self' refers to the instance of the class being created.
        """
        self.name = name  # Instance attribute
        self.age = age    # Instance attribute

    def bark(self):
        """An instance method defining a behavior of the Dog."""
        print(f"{self.name} says Woof! Woof!")

    def description(self):
        """An instance method returning a description of the dog."""
        return f"{self.name} is {self.age} years old."

# Creating objects (instances) of the Dog class
my_dog = Dog("Buddy", 3)
your_dog = Dog("Lucy", 5)

# Accessing attributes
print(f"My dog's name: {my_dog.name}")
print(f"Your dog's age: {your_dog.age}")
print(f"My dog's species: {my_dog.species}")
print(f"Your dog's species: {your_dog.species}") # Both share the class attribute

# Calling methods
my_dog.bark()        # Output: Buddy says Woof! Woof!
print(your_dog.description()) # Output: Lucy is 5 years old.
```
