```python
name = str(input("Enter String").lower())
vowels = ['a', 'e', 'i', 'o', 'u']
count = 0
for i in name:
    if i in vowels:
        count += 1
print("Number of vowels in the string is: ", count)
```

## Nested Loops

```python
for i in range(1,6):
    print(i*"* ")
```

```python
for i in range(1,6):
    for j in range(0,i):
        print("*", end=" ")
    print("")
```

```python
for i in range(1,6,+1):
    print(" "*(6-i), end="")
    for j in range(1,i+1):
        print(j, end=" ")

    print("")
```

```python
number = 1
for i in range(1,6,+1):
    for j in range(i):
        print(number, end=" ")
        number += 1
    print("")
```

```python
# fibonaki serie
number = int(input("Enter the number of terms: "))
a, b = 0, 1
print("Fibonacci sequence:")
for i in range(number):
    print(a, end=" ")
    a, b = b, a + b

print("")
```
