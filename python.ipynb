```python
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b

# Using the generator
for num in fibonacci(10):
    print(num)
```

```python
def fibonacci_iterative(n):
    fib_series = [0, 1]
    for i in range(2, n):
        fib_series.append(fib_series[i-1] + fib_series[i-2])
    return fib_series[:n]

# Example: Print first 10 Fibonacci numbers
print(fibonacci_iterative(10))
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

# Example usage
num = int(input("Enter a number: "))
if is_prime(num):
    print(f"{num} is a Prime Number")
else:
    print(f"{num} is NOT a Prime Number")
```

```python
text = "Programming"
print(text[0:4])     # 'Prog' (0 to 3)
print(text[:6])      # 'Progra' (start to index 5)
print(text[4:])      # 'ramming' (index 4 to end)
print(text[::2])     # 'Pormig' (every 2nd character)
print(text[::-1])    # 'gnimmargorP' (reversed string)
```
