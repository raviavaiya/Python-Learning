# Revrse string representation

```python
name="Ravi"
reversed_name=""
for char in name:
    reversed_name=char+reversed_name
print(reversed_name)
```

```python
reversed_name=name[::-1]
print(reversed_name)
```

```python
reversed_name = ''.join(reversed(name))
print(reversed_name)
```

# pattern Presentation

```python
for i in range(1,6+1):
    print(i*"* ")
```

```python
rows=6
for i in range(1,rows,+1):
    print("* "*i)
```

```python
for i in range(rows,0,-1):
    print("* "*i)
```

```python
rows = 5
for i in range(1, rows + 1):
    print(" " * (rows - i) + "* " * i)
```

```python
n = 5
for i in range(1, n + 1):
    print(" " * (n - i), end="")
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```
