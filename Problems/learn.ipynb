```python
import pandas as pd

# Sample employee data
data = {
    'Employee': ['Alice', 'Bob', 'Charlie', 'Diana', 'Ethan'],
    'Base Salary': [60000, 75000, 50000, 82000, 90000],
    'Performance Rating': [4.5, 3.8, 4.9, 3.2, 4.0]
}

# Create DataFrame
df = pd.DataFrame(data)

# Bonus calculation logic (example: 10% to 20% based on rating)
def calculate_bonus(salary, rating):
    if rating >= 4.5:
        return salary * 0.20
    elif rating >= 4.0:
        return salary * 0.15
    elif rating >= 3.5:
        return salary * 0.10
    else:
        return salary * 0.05

# Apply bonus calculation
df['Bonus'] = df.apply(lambda row: calculate_bonus(row['Base Salary'], row['Performance Rating']), axis=1)

# Total salary after bonus
df['Total Salary'] = df['Base Salary'] + df['Bonus']

print(df)
```

```python
import pandas as pd

# Sample data
data = {
    'Order ID': [101, 102, 103, 104, 105, 106, 102, 104],  # duplicate Order IDs
    'Category': ['Electronics', 'Clothing', 'Electronics', 'Books', 'Books', 'Clothing', 'Clothing', 'Books'],
    'Amount': [250, 80, 300, 20, 25, 75, 80, 20],
    'Quantity': [1, 2, 1, 3, 2, 1, 2, 3]
}

df = pd.DataFrame(data)

# Aggregate by category
category_summary = df.groupby('Category').agg({
    'Amount': 'sum',
    'Quantity': 'sum'
}).reset_index()

# Sort in descending order of total Amount
category_summary = category_summary.sort_values(by='Amount', ascending=False)

print("Aggregated and Sorted by Amount (Descending):")
category_summary=pd.DataFrame(category_summary)
print(category_summary)
```

```python
import pandas as pd

# Sample DataFrame
data = {
    'Department': ['Sales', 'Sales', 'HR', 'HR', 'IT', 'IT'],
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank'],
    'Salary': [50000, 55000, 60000, 62000, 70000, 72000]
}

df = pd.DataFrame(data)

# Group by department and calculate average salary
df.groupby('Department')['Salary'].mean()
```

```python
df['Region'] = ['East', 'West', 'East', 'West', 'East', 'West']
```

```python
# Group by Department and Region
df.groupby(['Department', 'Region'])['Salary'].mean()
```

```python
# Transform: Assign each row the average salary of their department
df['Dept_Avg_Salary'] = df.groupby('Department')['Salary'].transform('mean')
df
```
